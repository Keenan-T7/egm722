```python
import pandas as pd
import geopandas as gpd
import folium  # Imports the packages required

DEA2012 = gpd.read_file('data_files/DEA2012_2.shp') # Loads the shapefile for District Electoral Areas (2012)

df = pd.read_csv('data_files/emergencydepartments.csv') # Loads the csv file into a dataframe containing information about the 9 Emergency Departments in Northern Ireland

eds = gpd.GeoDataFrame(df[['Name','Address','Town', 'Postcode','Phone', 'Trust', 'Website']],
                       geometry=gpd.points_from_xy(df['X'],df['Y']),
                       crs='epsg:29903')  # Creates a GeoDataFrame from the emergency departments dataframe, telling it how to plot the points on the map and what projection to use (TM75 Irish Grid)

dea_travel = pd.read_csv('data_files/DEA2012.csv') # Loads the csv file for DEA2012 which contains information on the distance from each DEA area's centrepoint to its nearest Emergency Department

dea_merged = DEA2012.merge(dea_travel, left_on='DEACode', right_on='DEACode') # Merges the DEA Shapefile with the DEA CSV file

m = dea_merged.explore('RouteLength_km',
                   cmap='winter',
                   legend_kwds={'caption': 'Distance to nearest Emergency Department (km)'}
                  ) # shows the merged DEA data on the map, with each area shaded according to the distance from its centrepoint to its nearest Emergency Department

eds_args = {
    'm': m,
    'marker_type': 'marker',
    'popup': True,
    'legend': False,
    'marker_kwds': {'icon': folium.Icon(color='blue', icon='square-h', prefix='fa')}
}  # determines how the markers for the Emergency Departments will appear on the map

eds.explore('Name', **eds_args) # plots the updated markers on the map

m.save('DEA_EmgDept_Map.html') # saves the completed map to an html file
```
